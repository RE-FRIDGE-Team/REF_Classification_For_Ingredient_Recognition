```python
import mysql.connector
import json
import os

def export_household_to_json(db_config, save_path):
    """
    product_list 테이블의 생활용품 컬럼 데이터를 JSON 파일로 저장합니다.
    """
    try:
        conn = mysql.connector.connect(**db_config)
        cursor = conn.cursor()

        # 생활용품 컬럼에서 중복 없는 데이터 추출
        query = """
        SELECT DISTINCT 헬스_건강식품 
        FROM product_list 
        WHERE 헬스_건강식품 IS NOT NULL
        """
        cursor.execute(query)
        rows = cursor.fetchall()
        
        # 리스트로 변환
        household_items = [row[0] for row in rows]

        # 저장할 폴더가 없으면 생성
        dir_name = os.path.dirname(save_path)
        if dir_name and not os.path.exists(dir_name):
            os.makedirs(dir_name, exist_ok=True)

        # JSON 파일 저장
        with open(save_path, 'w', encoding='utf-8') as f:
            json.dump(household_items, f, ensure_ascii=False, indent=4)
        
        print(f"파일 저장 완료: {save_path}")
        print(f"총 {len(household_items)}개의 데이터가 저장되었습니다.")

    except Exception as e:
        print(f"오류 발생: {e}")
    finally:
        if 'conn' in locals() and conn.is_connected():
            cursor.close()
            conn.close()

if __name__ == "__main__":
    db_settings = {
        'host': '127.0.0.1',
        'user': 'inho123',
        'password': 'a3409440@', 
        'database': 'product_list_collection'
    }
    
    # 생활용품용 저장 경로와 파일명 설정
    target_path = r'C:/Users/LG/OneDrive/Desktop/RE_Fridge/data_collection/product_list/not_grocery_json/health_list.json'
    
    export_household_to_json(db_settings, target_path)
```

```python
import mysql.connector
import json
import os

def export_grocery_in_chunks(db_config, base_path, chunk_size=5000):
    try:
        # 데이터베이스 연결
        conn = mysql.connector.connect(**db_config)
        cursor = conn.cursor()

        # 특정 컬럼 데이터만 전체 조회
        query = "SELECT not_refined_grocery_name FROM grocery_list"
        cursor.execute(query)
        
        # 전체 데이터를 가져와서 리스트로 변환 (None 값 제외)
        rows = cursor.fetchall()
        all_names = [row[0] for row in rows if row[0] is not None]
        total_count = len(all_names)

        if total_count == 0:
            print("저장할 데이터가 없습니다.")
            return

        print(f"총 {total_count}개의 데이터를 추출했습니다. {chunk_size}개씩 분할 저장을 시작합니다.")

        # 데이터 분할 및 저장 로직
        for i in range(0, total_count, chunk_size):
            chunk = all_names[i : i + chunk_size]
            part_number = (i // chunk_size) + 1
            
            # 파일 경로 설정 (예: C:/data/grocery_part_1.json)
            file_name = f"grocery_part_{part_number}.json"
            full_path = os.path.join(base_path, file_name)

            # 해당 경로에 폴더가 없다면 생성
            if not os.path.exists(base_path):
                os.makedirs(base_path)

            with open(full_path, 'w', encoding='utf-8') as f:
                json.dump(chunk, f, ensure_ascii=False, indent=4)
            
            print(f"저장 완료: {full_path} ({len(chunk)}개)")

    except mysql.connector.Error as e:
        print(f"데이터베이스 에러: {e}")
    except Exception as e:
        print(f"에러 발생: {e}")
    finally:
        if 'conn' in locals() and conn.is_connected():
            cursor.close()
            conn.close()

if __name__ == "__main__":
    # 1. 데이터베이스 설정
    db_settings = {
        'host': '127.0.0.1',
        'user': 'inho123',
        'password': 'a3409440@',
        'database': 'product_list_collection'
    }
    
    # 2. 저장하고 싶은 폴더 경로를 여기에 입력하세요.
    # 예: 'C:/my_data/json_export' 또는 'D:/grocery_files'
    target_directory = 'C:/Users/LG/OneDrive/Desktop/RE_Fridge/data_collection/product_list/grocery_json_parts'
    
    # 실행 (5000개씩 분할 저장)
    export_grocery_in_chunks(db_settings, target_directory, chunk_size=5000)
```

```python
import time
import random
import mysql.connector
import undetected_chromedriver as uc
from bs4 import BeautifulSoup

class REF_product_category_collector:
	def __init__(self, db_config):
		self.db_config = db_config
		self.driver = None

	def initialize_driver(self):
		options = uc.ChromeOptions()
		options.add_argument("--headless")
		try:
			self.driver = uc.Chrome(options=options, version_main=144)
		except Exception:
			self.driver = uc.Chrome(options=options)

	def get_or_create_large_id(self, large_name):
		conn = mysql.connector.connect(**self.db_config)
		cursor = conn.cursor()
		try:
			cursor.execute("INSERT IGNORE INTO ref_category_large (large_name) VALUES (%s)", (large_name,))
			conn.commit()
			cursor.execute("SELECT large_id FROM ref_category_large WHERE large_name = %s", (large_name,))
			result = cursor.fetchone()
			return result[0] if result else None
		finally:
			cursor.close()
			conn.close()

	def insert_medium_category(self, large_id, medium_name):
		conn = mysql.connector.connect(**self.db_config)
		cursor = conn.cursor()
		try:
			query = "INSERT IGNORE INTO ref_category_medium (large_id, medium_name) VALUES (%s, %s)"
			cursor.execute(query, (large_id, medium_name))
			conn.commit()
			return True
		except:
			return False
		finally:
			cursor.close()
			conn.close()

	def stream_scrape(self, url_list):
		self.initialize_driver()
		
		# 필터링할 노이즈 키워드 정의
		noise_keywords = ["베스트", "세일", "신상", "특가", "혜택", "패션", "리빙", "전체보기", "필터"]
		
		for url in url_list:
			url = url.strip()
			if not url: continue
				
			yield f"\n[Network] Scanning: {url}"
			try:
				self.driver.get(url)
				time.sleep(6) # 렌더링 대기
				
				soup = BeautifulSoup(self.driver.page_source, 'html.parser')
				
				# 1. 대분류 정밀 추출 (페이지 상단 큰 제목 h2)
				# 클래스명이 복잡하므로 텍스트가 포함된 h2 요소를 직접 탐색
				large_name_tag = soup.find("h2")
				large_name = large_name_tag.get_text().strip() if large_name_tag else "미분류"
				
				# 만약 h2가 '필터'라면 다른 요소를 탐색 (Breadcrumb 영역)
				if any(nk in large_name for nk in ["필터", "카테고리"]):
					alternative_title = soup.select_one("div[class*='css-'] > h3")
					if alternative_title:
						large_name = alternative_title.get_text().strip()

				large_id = self.get_or_create_large_id(large_name)
				yield f"[Target] Large Category: {large_name} (ID: {large_id})"

				# 2. 중분류 정밀 추출 (사이드바 또는 상단 카테고리 필터 박스)
				# '전체보기'가 포함된 리스트 영역을 찾아 실제 상품군 텍스트만 추출
				category_list_container = soup.select_one("ul[class*='css-158p1a'], div[class*='CategoryFilter']")
				
				if category_list_container:
					items = category_list_container.find_all("a")
					for item in items:
						medium_name = item.get_text().strip()
						
						# 노이즈 및 전체보기 제외 로직
						if not medium_name or any(nk in medium_name for nk in noise_keywords):
							continue
							
						if self.insert_medium_category(large_id, medium_name):
							yield f"[Success] {large_name} > {medium_name} 저장 완료"
				else:
					yield "[Warning] 카테고리 영역을 찾지 못했습니다. 선택자를 재점검합니다."
				
			except Exception as e:
				yield f"[Error] {e}"

		if self.driver:
			self.driver.quit()

if __name__ == "__main__":
	db_info = {'host': '127.0.0.1',
               'user': 'inho123',
               'password': 'a3409440@',
               'database': 'product_list_collection'
              }
    
	collector = REF_product_category_collector(db_info)
	
	urls = input("Enter URLs: ").split(',')
	for message in collector.stream_scrape(urls):
		print(message)
```

```python
import time
import random
import mysql.connector
import undetected_chromedriver as uc
from bs4 import BeautifulSoup
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By

class REF_product_category_collector:
	def __init__(self, db_config):
		self.db_config = db_config
		self.driver = None

	def initialize_driver(self):
		"""브라우저가 닫혀있으면 새로 시작함"""
		if self.driver:
			try:
				self.driver.current_window_handle
				return # 이미 브라우저가 살아있으면 유지
			except:
				self.driver.quit()
		
		options = uc.ChromeOptions()
		options.add_argument("--headless")
		options.add_argument("--window-size=1920,1080")
		try:
			self.driver = uc.Chrome(options=options, version_main=144)
		except Exception:
			self.driver = uc.Chrome(options=options)

	def get_or_create_large_id(self, large_name):
		conn = mysql.connector.connect(**self.db_config)
		cursor = conn.cursor()
		try:
			cursor.execute("INSERT IGNORE INTO ref_category_large (large_name) VALUES (%s)", (large_name,))
			conn.commit()
			cursor.execute("SELECT large_id FROM ref_category_large WHERE large_name = %s", (large_name,))
			result = cursor.fetchone()
			return result[0] if result else None
		finally:
			cursor.close()
			conn.close()

	def insert_medium_category(self, large_id, medium_name):
		conn = mysql.connector.connect(**self.db_config)
		cursor = conn.cursor()
		try:
			query = "INSERT IGNORE INTO ref_category_medium (large_id, medium_name) VALUES (%s, %s)"
			cursor.execute(query, (large_id, medium_name))
			conn.commit()
			return True
		except:
			return False
		finally:
			cursor.close()
			conn.close()

	def stream_scrape(self, url_list):
		ignore_list = ["베스트", "세일", "패션", "리빙", "신상", "특가/혜택", "전체보기", "필터", "카테고리"]
		
		for url in url_list:
			url = url.strip()
			if not url: continue
			
			# 루프 시작 시 드라이버 상태 확인 및 재시작
			self.initialize_driver()
			yield f"\n[Scanning] Connecting to: {url}"
			
			try:
				self.driver.get(url)
				
				# 명시적 대기: 실제 카테고리 본문이 뜰 때까지 대기 (최대 15초)
				wait = WebDriverWait(self.driver, 15)
				wait.until(EC.presence_of_element_located((By.TAG_NAME, "h2")))
				time.sleep(random.uniform(3, 5)) # 안정적인 렌더링을 위한 추가 지연

				soup = BeautifulSoup(self.driver.page_source, 'html.parser')
				
				# 1. 대분류 추출 (제목)
				large_name = "미분류"
				title_tag = soup.find("h2")
				if title_tag:
					large_name = title_tag.get_text().strip()
				
				# 노이즈 체크
				if any(x in large_name for x in ["필터", "카테고리", "Unknown"]):
					alt_title = soup.select_one("h3")
					if alt_title: large_name = alt_title.get_text().strip()

				large_id = self.get_or_create_large_id(large_name)
				yield f"[Found] Large Category: {large_name} (ID: {large_id})"

				# 2. 중분류 추출 (이미지 속 박스 영역 타겟팅 강화)
				# 상세 페이지의 카테고리 박스 선택자
				content_box = soup.select_one("ul[class*='css-'], div[class*='CategoryFilter']")
				
				if content_box:
					links = content_box.find_all("a")
					for link in links:
						name = link.get_text().strip()
						if not name or any(ignore in name for ignore in ignore_list):
							continue
							
						if self.insert_medium_category(large_id, name):
							yield f"[Success] {large_name} > {name} 저장됨"
				else:
					yield "[Warning] 카테고리 본문 영역(박스)을 찾지 못했습니다."
				
			except Exception as e:
				yield f"[Error] 세션 또는 네트워크 문제 발생: {e}"
				if self.driver: self.driver.quit() # 에러 시 다음 루프에서 재시작 유도

		if self.driver: self.driver.quit()

if __name__ == "__main__":
	db_info = {'host': '127.0.0.1', 'user': 'inho123', 'password': 'a3409440@', 'database': 'product_list_collection'}
	collector = REF_product_category_collector(db_info)
	
	urls = input("Enter URLs: ").split(',')
	for msg in collector.stream_scrape(urls):
		print(msg)
```

```python
import mysql.connector
import json
import os # 폴더 생성을 위한 라이브러리 추가

def export_db_to_json(db_config, target_directory, file_name):
    # 1. 만약 지정한 폴더가 없다면 자동으로 생성합니다.
    if not os.path.exists(target_directory):
        os.makedirs(target_directory)
        print(f"새로운 폴더를 생성했습니다: {target_directory}")

    # 전체 파일 경로 결합
    full_path = os.path.join(target_directory, file_name)

    try:
        conn = mysql.connector.connect(**db_config)
        cursor = conn.cursor(dictionary=True)

        query = """
            SELECT l.large_name, m.medium_name, m.medium_id, m.created_at
            FROM ref_category_large l
            JOIN ref_category_medium m ON l.large_id = m.large_id
            ORDER BY l.large_id, m.medium_id
        """
        cursor.execute(query)
        rows = cursor.fetchall()

        full_data = {}
        for row in rows:
            l_name = row['large_name']
            if l_name not in full_data:
                full_data[l_name] = []
            
            full_data[l_name].append({
                "id": row['medium_id'],
                "name": row['medium_name'],
                "created": str(row['created_at'])
            })

        # 2. 지정된 경로에 JSON 파일로 저장
        with open(full_path, 'w', encoding='utf-8') as f:
            json.dump(full_data, f, ensure_ascii=False, indent=2)
            
        print(f"--- 백업 성공! ---")
        print(f"저장 위치: {full_path}")

    except Exception as e:
        print(f"오류 발생: {e}")
    finally:
        if conn.is_connected():
            cursor.close()
            conn.close()

# --- 사용자 설정 구역 ---
# 사용자님이 원하는 경로를 여기에 입력하세요. 
# (예: Windows는 'C:/Users/사용자이름/Desktop/backup')
my_path = "C:/Users/LG/OneDrive/Desktop/RE_Fridge/data_collection/product_list" 

db_info = {
    'host': '127.0.0.1', 
    'user': 'inho123', 
    'password': 'a3409440@', 
    'database': 'product_list_collection'
}

export_db_to_json(db_info, my_path, 'refridge_full_backup.json')
```

```python
import undetected_chromedriver as uc
from selenium.webdriver.common.by import By
import mysql.connector
import time
import random
import re

def build_cleaned_raw_collector(db_config, categories):
    
    options = uc.ChromeOptions()
    session_id = random.randint(1, 9999)
    options.add_argument(f'--user-data-dir=C:/temp/clean_raw_{session_id}')
    
    driver = uc.Chrome(options=options, version_main=144)

    price_rx = re.compile(r'\d{1,3}(,\d{3})*\s?원')
    noise_rx = re.compile(r'로켓배송|무료배송|일시품절|새벽배송|내일\s?도착|최대\s?\d+%\s?적립|로켓프레시|정기배송|입점신청')
    spec_rx = re.compile(r'\d+(\.\d+)?\s?(ml|l|kg|g|개입|입|p|매|v|w|호|인용|단|팩|장|cm|mm|m)')

    try:
        conn = mysql.connector.connect(**db_config)
        cursor = conn.cursor()

        for cat_id in categories:
            print(f"\n[Process] Category {cat_id} 제품명 정밀 추출 시작")
            
            for page in range(1, 10):
                url = f"https://www.coupang.com/np/categories/{cat_id}?page={page}"
                driver.get(url)
                
                time.sleep(random.uniform(5, 7))
                driver.execute_script("window.scrollTo(0, 2000);")
                time.sleep(2)

                extract_script = """
                var results = [];
                var nodes = document.querySelectorAll('li, div[class*="item"], dt, a');
                for (var i = 0; i < nodes.length; i++) {
                    var text = nodes[i].innerText || nodes[i].textContent;
                    if (text && text.length > 10) {
                        results.push(text.trim().replace(/\\n/g, " "));
                    }
                }
                return results;
                """
                raw_texts = driver.execute_script(extract_script)

                cleaned_names_set = set()
                for text in raw_texts:
                    if "원" in text and ("배송" in text or "도착" in text or "로켓" in text):
                        cleaned = price_rx.sub('', text)
                        cleaned = noise_rx.sub('', cleaned)
                        cleaned = spec_rx.sub('', cleaned)
                        cleaned = re.sub(r'[\(\)\[\]\+\-\/\*]', ' ', cleaned)
                        cleaned = ' '.join(cleaned.split())
                        
                        if len(cleaned) > 2:
                            cleaned_names_set.add(cleaned[:255])

                if cleaned_names_set:
                    # 'product_list' 테이블의 '패션의류_잡화' 컬럼에 저장
                    sql = "INSERT IGNORE INTO product_list (헬스_건강식품) VALUES (%s)"
                    cursor.executemany(sql, [(name,) for name in cleaned_names_set])
                    conn.commit()
                    
                    # 수집된 리스트 중 마지막 아이템 확인
                    last_data = list(cleaned_names_set)[-1]
                    
                    print(f"   (Page {page}) {len(cleaned_names_set)}개 제품명 저장 완료")
                    print(f"   마지막 데이터 : {last_data}")
                else:
                    print(f"   (Page {page}) 유효 데이터 없음")

                time.sleep(random.uniform(5, 7))

    except Exception as e:
        print(f"에러 발생: {e}")
    finally:
        if 'conn' in locals() and conn.is_connected():
            cursor.close()
            conn.close()

if __name__ == "__main__":
    db_settings = {
        'host': '127.0.0.1',
        'user': 'inho123',
        'password': 'a3409440@', 
        'database': 'product_list_collection'
    }
    
    # 카테고리 ID
    target_ids = ['501452','334193','310885']
    
    build_cleaned_raw_collector(db_settings, target_ids)
```
